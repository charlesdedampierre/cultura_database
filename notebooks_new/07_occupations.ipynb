# Occupations — share of active individuals over time (World, Roman Empire, Tang Dynasty) and astronomer co-occupation networks

```python
import random
import tempfile
from collections import Counter
from itertools import combinations

import duckdb
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import polars as pl

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

DB_PATH = '../data/cultura/humans_clean_v2.duckdb'

COLORS = {
    'astronomer': '#6f9fca',
    'astrologer': '#f4b6b0',
    'writer':     '#b5dbb0',
    'other_node': '#e3e9f2',
    'edge':       '#6b7280',
    'Other':      '#dddddd',
}

plt.rcParams.update({
    'figure.dpi':        120,
    'figure.figsize':    (16, 7.5),
    'axes.spines.top':   False,
    'axes.spines.right': False,
    'font.family':       'DejaVu Sans',
    'axes.labelsize':    18,
    'xtick.labelsize':   15,
    'ytick.labelsize':   15,
    'legend.fontsize':   13,
})
```

## Parameters

An individual is *active* during their peak-productivity window (`individual_enriched.peak_productivity`,
start/end years). This is the DB's implementation of the impact-year rule (productive-age window 29–53 anchored
on birth, or floruit / works years when available). Time figures use 50-year buckets (mean number of active
individuals per year in the bucket); a trailing incomplete bucket is dropped.

```python
WORLD_RANGE = (-500, 2018)
POLITY_NAMES = ['Roman Empire', 'Tang Dynasty']

TOP_OCCUPATIONS_PER_CENTURY = 10
MAX_OCCUPATIONS_PER_SCOPE = 20
BUCKET_YEARS = 50

ASTRONOMER_PERIODS = {
    '750 – 1250':  (750, 1250),
    '1400 – 1650': (1400, 1650),
    '1750 – 1950': (1750, 1950),
}
ASTRONOMER_VARIANTS = ['amateur astronomer', 'radio astronomer', 'archaeoastronomer', 'astronomer royal']

NETWORK_NODE_COUNT = 15
NETWORK_MIN_EDGE_COUNT = 2
```

## Load data

`occupation_rows` = one row per (individual, occupation label) with the peak-productivity window.
Occupation labels come from the `occupation` table (`individual.occupation` only holds qids).
Several occupation qids share one English label (e.g. three items labelled *researcher*); like the old
notebook we group by label, and a person is counted once per label.

```python
con = duckdb.connect(DB_PATH, read_only=True)
con.sql(f"SET memory_limit='3GB'; SET threads=4; SET temp_directory='{tempfile.mkdtemp()}';")

OCCUPATION_ROWS_SQL = """
    WITH people AS (
        SELECT entity.qid                   AS qid,
               peak_productivity.start_year AS peak_start,
               peak_productivity.end_year   AS peak_end
        FROM   individual_enriched
        WHERE  peak_productivity.start_year IS NOT NULL {cohort_filter}
    ),
    occupation_qids AS (
        SELECT entity.qid AS qid, unnest(occupation).entity.qid AS occupation_qid
        FROM   individual
        WHERE  len(occupation) > 0
    ),
    occupation_rows AS (
        SELECT DISTINCT people.qid, people.peak_start, people.peak_end, o.entity.label_en AS occupation
        FROM   people
        JOIN   occupation_qids USING (qid)
        JOIN   occupation AS o ON o.entity.qid = occupation_qids.occupation_qid
        WHERE  o.entity.label_en IS NOT NULL
    )
"""
WORLD_ROWS = OCCUPATION_ROWS_SQL.format(cohort_filter='')

def polity_rows_sql(polity_name):
    return OCCUPATION_ROWS_SQL.format(
        cohort_filter=f"AND list_contains(list_transform(polity, x -> x.polity.name), '{polity_name}')")

world_overview = con.sql(WORLD_ROWS + """
    SELECT COUNT(*) AS occupation_rows, COUNT(DISTINCT qid) AS individuals,
           MIN(peak_start) AS min_start, MAX(peak_end) AS max_end
    FROM   occupation_rows
""").pl()
world_overview
```

### Polity year spans (from `polity.territories`), rounded outward to the century

```python
polity_year_spans = con.sql("""
    SELECT name AS polity_name, MIN(t.start_year) AS polity_start, MAX(t.end_year) AS polity_end
    FROM   (SELECT name, unnest(territories) AS t FROM polity WHERE list_contains($names, name))
    GROUP  BY name
""", params={'names': POLITY_NAMES}).pl()

assert polity_year_spans.height == len(POLITY_NAMES)
polity_to_year_range = {
    row['polity_name']: ((row['polity_start'] // 100) * 100, ((row['polity_end'] + 99) // 100) * 100 - 1)
    for row in polity_year_spans.iter_rows(named=True)
}
ROMAN_RANGE = polity_to_year_range['Roman Empire']
TANG_RANGE = polity_to_year_range['Tang Dynasty']
print('ROMAN_RANGE:', ROMAN_RANGE, ' TANG_RANGE:', TANG_RANGE)
polity_year_spans
```

### Polity cohorts (individuals whose `individual_enriched.polity` list contains the polity)

```python
occupation_rows_roman = con.sql(polity_rows_sql('Roman Empire') + 'SELECT * FROM occupation_rows').pl()
occupation_rows_tang = con.sql(polity_rows_sql('Tang Dynasty') + 'SELECT * FROM occupation_rows').pl()

for name, rows in [('Roman Empire', occupation_rows_roman), ('Tang Dynasty', occupation_rows_tang)]:
    print(f'{name:13s}: {rows.height:,} occupation rows, {rows["qid"].n_unique():,} individuals')
occupation_rows_tang.head()
```

### Astronomers

Individuals with at least one occupation label containing *astronomer*, with all their occupation labels.

```python
astronomer_individuals = con.sql(WORLD_ROWS + """
    SELECT qid, peak_start, list(occupation ORDER BY occupation) AS occupations
    FROM   occupation_rows
    GROUP  BY qid, peak_start
    HAVING bool_or(occupation LIKE '%astronomer%')
""").pl()

assert astronomer_individuals['qid'].is_unique
print('shape:', astronomer_individuals.shape)
astronomer_individuals.head()
```

## Preprocessing

### Top occupations per scope

Visible extraction rule: within the scope's year range, individuals are bucketed by the century of their
peak-window start; occupations are ranked per century; an occupation is kept if it enters the top
10 of at least one century; the kept ones are ordered by their summed counts and capped at 20.

```python
TOP_SQL = """
    , century_counts AS (
        SELECT CAST(floor(peak_start / 100) * 100 AS BIGINT) AS century, occupation, COUNT(*) AS n
        FROM   occupation_rows
        WHERE  peak_start BETWEEN {lo} AND {hi}
        GROUP  BY ALL
    ),
    ranked AS (
        SELECT *, row_number() OVER (PARTITION BY century ORDER BY n DESC, occupation) AS rank_in_century
        FROM   century_counts
    )
    SELECT occupation, CAST(SUM(n) AS BIGINT) AS total_individuals, MIN(rank_in_century) AS best_rank
    FROM   ranked
    WHERE  rank_in_century <= {top}
    GROUP  BY occupation
    ORDER  BY total_individuals DESC, best_rank, occupation
    LIMIT  {limit}
"""

def top_occupations(rows_sql, year_range):
    lo, hi = year_range
    return con.sql(rows_sql + TOP_SQL.format(lo=lo, hi=hi, top=TOP_OCCUPATIONS_PER_CENTURY,
                                             limit=MAX_OCCUPATIONS_PER_SCOPE)).pl()

world_top = top_occupations(WORLD_ROWS, WORLD_RANGE)
roman_top = top_occupations(polity_rows_sql('Roman Empire'), ROMAN_RANGE)
tang_top = top_occupations(polity_rows_sql('Tang Dynasty'), TANG_RANGE)

for name, top in [('world', world_top), ('roman', roman_top), ('tang', tang_top)]:
    assert top['occupation'].is_unique
    print(f'{name}: {top.height} occupations')
world_top.head(10)
```

### Active individuals per year, per occupation group

For each occupation group (top occupations + *Other*), the peak window is clipped to the scope range and
the number of active individuals per year is obtained from +1 / −1 events (a difference array).

```python
ACTIVE_SQL = """
    , grouped AS (
        SELECT CASE WHEN list_contains($top, occupation) THEN occupation ELSE 'Other' END AS occupation_group,
               greatest(peak_start, {lo}) AS s,
               least(peak_end, {hi})      AS e
        FROM   occupation_rows
    )
    SELECT occupation_group, year, SUM(delta) AS delta
    FROM   (SELECT occupation_group, s AS year, 1 AS delta FROM grouped WHERE s <= e
            UNION ALL
            SELECT occupation_group, e + 1 AS year, -1 AS delta FROM grouped WHERE s <= e)
    GROUP  BY ALL
"""

def active_per_year(rows_sql, top, year_range):
    lo, hi = year_range
    order = top['occupation'].to_list() + ['Other']
    deltas = con.sql(rows_sql + ACTIVE_SQL.format(lo=lo, hi=hi), params={'top': order[:-1]}).pl()
    years = np.arange(lo, hi + 2)
    counts = {}
    for occupation in order:
        d = deltas.filter(pl.col('occupation_group') == occupation)
        arr = np.zeros(len(years), dtype=np.int64)
        np.add.at(arr, d['year'].to_numpy() - lo, d['delta'].to_numpy())
        counts[occupation] = np.cumsum(arr)[:-1]
    return pl.DataFrame({'year': years[:-1], **counts}), order

world_active, world_order = active_per_year(WORLD_ROWS, world_top, WORLD_RANGE)
roman_active, roman_order = active_per_year(polity_rows_sql('Roman Empire'), roman_top, ROMAN_RANGE)
tang_active, tang_order = active_per_year(polity_rows_sql('Tang Dynasty'), tang_top, TANG_RANGE)

con.close()

assert (world_active.drop('year').to_numpy() >= 0).all()
print('world shape:', world_active.shape)
world_active.head()
```

### 50-year buckets and shares

Mean yearly active count per bucket, then share of each occupation group in the bucket. The trailing bucket is
dropped when the scope range does not cover it entirely.

```python
def bucket_shares(active, order):
    lo, hi = active['year'].min(), active['year'].max()
    bucketed = (
        active
        .with_columns(((pl.col('year') - lo) // BUCKET_YEARS * BUCKET_YEARS + lo).alias('bucket'))
        .group_by('bucket').agg(pl.len().alias('n_years'), *[pl.col(o).mean() for o in order])
        .sort('bucket')
    )
    complete = bucketed.filter(pl.col('n_years') == BUCKET_YEARS)
    print(f'buckets: {bucketed.height} -> {complete.height} complete ({complete["bucket"].min()}–{complete["bucket"].max() + BUCKET_YEARS - 1})')
    total = pl.sum_horizontal(order)
    return (complete.with_columns(total.alias('total'))
                    .filter(pl.col('total') > 0)
                    .with_columns([(pl.col(o) / pl.col('total')).alias(o) for o in order]))

world_shares = bucket_shares(world_active, world_order)
roman_shares = bucket_shares(roman_active, roman_order)
tang_shares = bucket_shares(tang_active, tang_order)

assert np.allclose(world_shares.select(world_order).to_numpy().sum(axis=1), 1)
world_shares.head()
```

### Shared occupation palette

```python
all_occupations = list(dict.fromkeys(world_order[:-1] + roman_order[:-1] + tang_order[:-1]))
base_colors = list(plt.cm.tab20.colors) + list(plt.cm.tab20b.colors) + list(plt.cm.tab20c.colors)
occupation_palette = {o: base_colors[i % len(base_colors)] for i, o in enumerate(all_occupations)}
occupation_palette['Other'] = COLORS['Other']
print(f'shared palette: {len(all_occupations)} occupations + Other')
```

```python
def format_year(year):
    year = int(round(year))
    return f'{abs(year)} BCE' if year < 0 else f'{year}'


def plot_share_area(shares, order, scope_name, log_time=False, ref_year=2020, tick_years=None):
    mids = shares['bucket'].to_numpy() + BUCKET_YEARS / 2
    x = -np.log10(ref_year - mids) if log_time else mids
    if tick_years is None:
        tick_years = shares['bucket'].to_list()
    ticks = [-np.log10(ref_year - y) for y in tick_years] if log_time else tick_years

    fig, ax = plt.subplots()
    ax.stackplot(x, shares.select(order).to_numpy().T, labels=order,
                 colors=[occupation_palette[o] for o in order], edgecolor='white', linewidth=0.2, alpha=0.9)
    ax.set_xticks(ticks)
    ax.set_xticklabels([format_year(y) for y in tick_years], rotation=30, ha='right')
    ax.set_xlim(x.min(), x.max())
    ax.set_ylim(0, 1)
    ax.set_xlabel('Year (log time)' if log_time else 'Year')
    fig.suptitle(scope_name, fontsize=26, color='black', y=0.98)
    handles, labels = ax.get_legend_handles_labels()
    ax.legend(handles[::-1], labels[::-1], frameon=False, loc='upper left', bbox_to_anchor=(1.01, 1.0),
              borderaxespad=0, handlelength=1.2, labelspacing=0.4)
    fig.subplots_adjust(left=0.06, right=0.78, top=0.91, bottom=0.16)
    plt.show()
```

## Astronomer co-occupations

### Cohort per period (by peak-window start)

```python
period_of = pl.lit(None, dtype=pl.Utf8)
for label, (lo, hi) in reversed(ASTRONOMER_PERIODS.items()):
    period_of = pl.when(pl.col('peak_start').is_between(lo, hi)).then(pl.lit(label)).otherwise(period_of)

astronomers_with_period = astronomer_individuals.with_columns(period_of.alias('period')).filter(pl.col('period').is_not_null())
print(f'astronomers: {astronomer_individuals.height:,} -> {astronomers_with_period.height:,} within the three periods')

astronomers_per_period = astronomers_with_period.group_by('period').len().rename({'len': 'astronomer_count'}).sort('period')
astronomers_per_period
```

```python
occupation_counts_per_period = {label: Counter() for label in ASTRONOMER_PERIODS}
pair_counts_per_period = {label: Counter() for label in ASTRONOMER_PERIODS}

for row in astronomers_with_period.iter_rows(named=True):
    occupations = [o for o in row['occupations'] if o.lower() not in ASTRONOMER_VARIANTS]
    occupation_counts_per_period[row['period']].update(occupations)
    pair_counts_per_period[row['period']].update(combinations(sorted(occupations), 2))

top_cooccupations = pl.concat([
    pl.DataFrame(occupation_counts_per_period[p].most_common(NETWORK_NODE_COUNT),
                 schema=['occupation', 'individual_count'], orient='row').with_columns(pl.lit(p).alias('period'))
    for p in ASTRONOMER_PERIODS
]).select('period', 'occupation', 'individual_count')

for p in ASTRONOMER_PERIODS:
    print(f'{p:13s}: {len(occupation_counts_per_period[p]):4d} unique occupations, {len(pair_counts_per_period[p]):5d} unique pairs')
top_cooccupations.head(10)
```

### Fig 4 — Astronomer co-occupation networks

```python
def node_color(occupation):
    return COLORS[occupation] if occupation in ('astronomer', 'astrologer', 'writer') else COLORS['other_node']

fig, axes = plt.subplots(1, 3, figsize=(22, 8), dpi=150)
for ax, period in zip(axes, ASTRONOMER_PERIODS):
    counts = occupation_counts_per_period[period]
    top = [o for o, _ in counts.most_common(NETWORK_NODE_COUNT)]
    if 'astronomer' not in top:
        top = ['astronomer'] + top[:NETWORK_NODE_COUNT - 1]

    graph = nx.Graph()
    graph.add_nodes_from((o, {'count': counts[o]}) for o in top)
    graph.add_edges_from((a, b, {'weight': w}) for (a, b), w in pair_counts_per_period[period].items()
                         if a in graph and b in graph and w >= NETWORK_MIN_EDGE_COUNT)
    layout = nx.spring_layout(graph, k=2.0, iterations=150, seed=SEED, weight='weight')

    max_count = max(nx.get_node_attributes(graph, 'count').values())
    weights = [graph.edges[e]['weight'] for e in graph.edges()]
    max_weight = max(weights) if weights else 1
    for edge, w in zip(graph.edges(), weights):
        nx.draw_networkx_edges(graph, layout, edgelist=[edge], ax=ax, width=0.5 + 4.5 * w / max_weight,
                               edge_color=COLORS['edge'], alpha=0.15 + 0.55 * w / max_weight)
    nx.draw_networkx_nodes(graph, layout, ax=ax, node_size=[400 + 2500 * graph.nodes[n]['count'] / max_count for n in graph],
                           node_color=[node_color(n) for n in graph], edgecolors='#555555', linewidths=0.8)
    nx.draw_networkx_labels(graph, layout, ax=ax, font_size=12)

    n = astronomers_per_period.filter(pl.col('period') == period)['astronomer_count'].item()
    ax.text(0.5, -0.04, f'{period}  (n = {n:,})', transform=ax.transAxes, ha='center', va='top', fontsize=16)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
