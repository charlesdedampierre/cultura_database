## External Identifier Types (Catalog Export)

This notebook exports the full list of external identifier types (catalogs) linked to individuals in the database.
The complete list is written to `tables/table_annex_identifiers.csv`, and the top 50 identifier types by count are displayed below.

```python
import sqlite3
import csv
import pandas as pd

DB = '../data/humans_clean.sqlite3'

conn = sqlite3.connect(DB)
cur = conn.execute("""
    SELECT property_id, name_en, count, description
    FROM identifier_types
    ORDER BY count DESC
""")
rows = cur.fetchall()
conn.close()

# Export CSV
csv_path = 'tables/table_annex_identifiers.csv'
with open(csv_path, 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['Property ID', 'Name', 'Individuals', 'Description'])
    for pid, name, count, desc in rows:
        writer.writerow([pid, name or '', count, desc or ''])

print(f'{len(rows)} identifier types exported to {csv_path}')

# Display first 50
df = pd.DataFrame(rows[:50], columns=['Property ID', 'Name', 'Individuals', 'Description'])
df['Individuals'] = df['Individuals'].apply(lambda x: f'{x:,}')
df['Description'] = df['Description'].fillna('').str[:80]
df.style.hide(axis='index')
```
